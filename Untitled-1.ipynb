```python
import pandas as pd
import numpy as np
sales = pd.read_csv("secs.csv")
prices = pd.read_csv("RetailerPrice.csv")
```

```python
# MAKE THE KEY OF SKUCODE | BATCHNAME

sales["key"] = (
    sales["SKU_CODE"].astype(str).str.strip()
    + "|"
    + sales["Batch_Name"].astype(str).str.strip()
)

prices["key"] = (
    prices["skucode1"].astype(str).str.strip()
    + "|"
    + prices["batch_name1"].astype(str).str.strip()
)
```

```python
# 52131 sales rows have a matching key in prices. i.e. same skucode|batchname exists in the suggested prices list. 
# 2482 sales rows do not have a matching key in prices.

sales["key"].isin(prices["key"]).value_counts()
```

```python
## checks for duplicate keys i.e. the skucode|batchname thing

prices["key"].duplicated().sum()      
```

```python
price_lookup = prices.set_index("key")  
price_lookup.head()
price_lookup.shape
```

```python
sales["Suggested_MRP"] = sales["key"].map(price_lookup["mrp1"])    #python lookup. take sales key, look for that key in suggested price table and paste it in the new column suggested mrp in the sales table
sales["Suggested_Unit_Basic_With_Tax"] = sales["key"].map(
    price_lookup["Basic with tax"]
)  
## same with suggested unit with tax
```

```python
## null values - whose keys arent found in prices.  

sales[["Suggested_MRP", "Suggested_Unit_Basic_With_Tax"]].isna().sum()
```

```python
### DATAFRAME UNSUGGESTED ITEM 
# Keep a record of all sales where suggested price could not be mapped


unsuggested_items = sales[sales["Suggested_MRP"].isna()].copy()

# Add reason/status
unsuggested_items["Price_Match_Status"] = "Suggested Price Not Available"

# Summary
print("Unsuggested rows:", len(unsuggested_items))
print(
    "Unsuggested billing:",
    unsuggested_items["Total_Invoice_Amount"].sum()
)

# View the records
unsuggested_items.head()
```

ANALYSIS. MASTER SHEET. NOW AFTER THIS I AM DROPPING THE NAN VALUES FROM THE DF

```python
df1 = analysis.dropna(
    subset=["Suggested_Unit_Basic_With_Tax"]
).copy()
```

```python
## index = false means that the first column won't have 0,1,2.. index etc. 
df1.to_csv("df1.csv", index=False)
```

```python
import numpy as np

df1["Price_Classification"] = np.where(     #create column 
    np.isclose(
        df1["secs_unitbasic"],
        df1["Suggested_Unit_Basic_With_Tax"],
        rtol=0,    
        atol=1      
    ),
    "At Suggested Price",
    np.where(
        df1["secs_unitbasic"] < df1["Suggested_Unit_Basic_With_Tax"],
        "Below Suggested Price",
        "Above Suggested Price"
    )
)
```

```python
df1["Price_Classification"].value_counts()
```

```python
df1.to_csv("df1_price_analysis.csv", index=False) 
```

```python
distributor_billing = (
    df1.groupby(
        ["Distributor ERP Code", "Price_Classification"]
    )["Total_Invoice_Amount"]
    .sum()
    .reset_index()
)
```

```python
distributor_billing.head(20)
```

```python
distributor_billing["Distributor_Total_Billing"] = (
    distributor_billing
    .groupby("Distributor ERP Code")["Total_Invoice_Amount"]
    .transform("sum")
)
```

```python
distributor_billing["Billing_Percentage"] = (
    distributor_billing["Total_Invoice_Amount"]
    / distributor_billing["Distributor_Total_Billing"]
    * 100
)
```

```python
distributor_billing.head(20)
```

```python
distributor_summary = distributor_billing.pivot(
    index="Distributor ERP Code",
    columns="Price_Classification",
    values="Billing_Percentage"
).reset_index()
```

```python
distributor_summary = distributor_summary.fillna(0)
```

```python
distributor_summary.head(20)
```

```python
distributor_summary["Distributor ERP Code"].nunique()
```

```python
print(
    "Distributors with Above Suggested billing:",
    (distributor_summary["Above Suggested Price"] > 0).sum()
)

print(
    "Distributors with Below Suggested billing:",
    (distributor_summary["Below Suggested Price"] > 0).sum()
)

print(
    "Distributors with 100% Suggested Price billing:",
    (distributor_summary["At Suggested Price"] == 100).sum()
)
```

```python
# Distributor-level pricing behaviour counts

total_distributors = len(distributor_summary)

fully_suggested = (
    distributor_summary["At Suggested Price"] >= 99.999
).sum()

mixed = (
    (distributor_summary["At Suggested Price"] > 0) &
    (distributor_summary["At Suggested Price"] < 99.999)
).sum()

no_suggested = (
    distributor_summary["At Suggested Price"] == 0
).sum()

any_above = (
    distributor_summary["Above Suggested Price"] > 0
).sum()

any_below = (
    distributor_summary["Below Suggested Price"] > 0
).sum()

print("Total distributors:", total_distributors)
print("100% at suggested price:", fully_suggested)
print("Mixed pricing:", mixed)
print("0% at suggested price:", no_suggested)
print("Distributors with any above-suggested billing:", any_above)
print("Distributors with any below-suggested billing:", any_below)
```

```python
import numpy as np

# Mark billing that is NOT at suggested price
df1["Non_Suggested_Billing"] = np.where(
    df1["Price_Classification"] != "At Suggested Price",
    df1["Total_Invoice_Amount"],
    0
)

# Distributor-wise summary
distributor_deviation = (
    df1.groupby("Distributor ERP Code")
    .agg(
        Total_Billing=("Total_Invoice_Amount", "sum"),
        Non_Suggested_Billing=("Non_Suggested_Billing", "sum")
    )
    .reset_index()
)

# Calculate non-suggested billing %
distributor_deviation["Non_Suggested_Pct"] = (
    distributor_deviation["Non_Suggested_Billing"]
    / distributor_deviation["Total_Billing"]
    * 100
)

distributor_deviation
```

```python
top_non_suggested = distributor_deviation.sort_values(
    "Non_Suggested_Billing",
    ascending=False
)
top_non_suggested.head(20)
```

```python
df1["Above_Suggested_Billing"] = np.where(
    df1["Price_Classification"] == "Above Suggested Price",
    df1["Total_Invoice_Amount"],
    0
)

df1["Below_Suggested_Billing"] = np.where(
    df1["Price_Classification"] == "Below Suggested Price",
    df1["Total_Invoice_Amount"],
    0
)

df1["At_Suggested_Billing"] = np.where(
    df1["Price_Classification"] == "At Suggested Price",
    df1["Total_Invoice_Amount"],
    0
)

distributor_detailed = (
    df1.groupby("Distributor ERP Code")
    .agg(
        Total_Billing=("Total_Invoice_Amount", "sum"),
        At_Suggested_Billing=("At_Suggested_Billing", "sum"),
        Above_Suggested_Billing=("Above_Suggested_Billing", "sum"),
        Below_Suggested_Billing=("Below_Suggested_Billing", "sum")
    )
    .reset_index()
)

distributor_detailed["At_Suggested_Pct"] = (
    distributor_detailed["At_Suggested_Billing"]
    / distributor_detailed["Total_Billing"] * 100
)

distributor_detailed["Above_Suggested_Pct"] = (
    distributor_detailed["Above_Suggested_Billing"]
    / distributor_detailed["Total_Billing"] * 100
)

distributor_detailed["Below_Suggested_Pct"] = (
    distributor_detailed["Below_Suggested_Billing"]
    / distributor_detailed["Total_Billing"] * 100
)

distributor_detailed.head(20)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(20, 4))

# -------------------------
# 1. Billing Split
# -------------------------
billing_chart = (
    df1.groupby("Price_Classification")["Total_Invoice_Amount"]
    .sum()
)

billing_chart = billing_chart / billing_chart.sum() * 100

order = [
    "At Suggested Price",
    "Above Suggested Price",
    "Below Suggested Price"
]

billing_chart = billing_chart.reindex(order)

axes[0].bar(
    ["At", "Above", "Below"],
    billing_chart.values,
    color=["#2E7D32", "#F59E0B", "#C62828"]
)

axes[0].set_title("Billing Split")
axes[0].set_ylabel("% of Billing")

for i, v in enumerate(billing_chart.values):
    axes[0].text(i, v + 1, f"{v:.1f}%", ha="center", fontsize=9)


# -------------------------
# 2. Distributor Behaviour
# -------------------------
labels = ["100% Suggested", "Mixed", "Any Above", "Any Below"]

values = [
    fully_suggested,
    mixed,
    any_above,
    any_below
]

axes[1].barh(
    labels,
    values,
    color=["#2E7D32", "#607D8B", "#F59E0B", "#C62828"]
)

axes[1].set_title("Distributor Behaviour")

for i, v in enumerate(values):
    axes[1].text(v + 2, i, str(v), va="center", fontsize=9)


# -------------------------
# 3. Top 10 by Non-Suggested ₹
# -------------------------
top10_value = (
    distributor_detailed
    .nlargest(10, "Non_Suggested_Billing")
    .sort_values("Non_Suggested_Billing")
    .copy()
)

top10_value["Non_Suggested_Lakh"] = (
    top10_value["Non_Suggested_Billing"] / 100000
)

axes[2].barh(
    top10_value["Distributor ERP Code"],
    top10_value["Non_Suggested_Lakh"],
    color="#7C3AED"
)

axes[2].set_title("Top 10: Non-Suggested ₹")
axes[2].set_xlabel("₹ Lakh")
axes[2].tick_params(axis="y", labelsize=7)


# -------------------------
# 4. Top 10 by Non-Suggested %
# -------------------------
top10_pct = (
    distributor_detailed
    .nlargest(10, "Non_Suggested_Pct")
    .sort_values("Non_Suggested_Pct")
)

axes[3].barh(
    top10_pct["Distributor ERP Code"],
    top10_pct["Non_Suggested_Pct"],
    color="#2563EB"
)

axes[3].set_title("Top 10: Non-Suggested %")
axes[3].set_xlabel("% of Billing")
axes[3].tick_params(axis="y", labelsize=7)

plt.tight_layout()
plt.show()
```
